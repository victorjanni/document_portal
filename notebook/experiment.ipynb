```python
from langchain_community.document_loaders import PyPDFLoader
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
import os
os.getcwd()
```

```python
file_path = os.path.join(os.getcwd(), "data", "sample.pdf")
```

```python
loader = PyPDFLoader(file_path)
```

```python
documents = loader.load()
```

```python
documents
```

```python
len(documents)
```

```python
text_splitter=RecursiveCharacterTextSplitter(
    chunk_size = 750,
    chunk_overlap =150,
    length_function = len
)
```

```python
docs=text_splitter.split_documents(documents)
```

```python
vectorstore = FAISS.from_documents(docs, embedding_model)
```

```python
relevant_doc = vectorstore.similarity_search("what is Distributed LLM Training")
```

```python
relevant_doc[0].page_content
```

```python
relevant_doc = vectorstore.similarity_search("what is Distributed LLM Training", k=5)
```

```python
relevant_doc[0].page_content
```

```python
relevant_doc[1].page_content
```

```python
retriever = vectorstore.as_retriever(search_kwargs={"k": 5})
```

```python
retriever.invoke("what is Distributed LLM Training")
```

```python
from langchain_core.prompts import PromptTemplate
```

```python
prompt_template = """
       Answer the question based on the context provide below.
       if the context does not contain sufficient information, respond with:
        "I do not have enough information about your query."
        
        context: {context}
        
        Question: {question}
        
        Answer: """
```

```python
prompt=PromptTemplate(
    template=prompt_template,
    input_variables=["context","question"]
)
```

```python
prompt
```

```python
from langchain_core.output_parsers import StrOutputParser
```

```python
parser = StrOutputParser()
```

```python
def format_docs(docs):
    return "\n\n".join([doc.page_content for doc in docs])
```

```python
from langchain_core.runnables import RunnablePassthrough
```

```python
rag_chain = (
    {"context": retriever | format_docs, "question": RunnablePassthrough()}
    | prompt
    | llm
    |StrOutputParser()
)
```

```python
rag_chain.invoke("what is Distributed LLM Training?")
```
